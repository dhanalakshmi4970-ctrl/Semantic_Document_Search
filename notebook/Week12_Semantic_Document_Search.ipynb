# Week 12 - Semantic Document Search
## LangChain and ChromaDB

This notebook demonstrates semantic document retrieval, keyword search and similarity score comparison.

```
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'dataset').exists():
    ROOT = ROOT.parent

results = pd.read_csv(ROOT / 'dataset' / 'search_results.csv')
scores = pd.read_csv(ROOT / 'dataset' / 'similarity_scores.csv')

print('Total search results:', len(results))
print('Total similarity scores:', len(scores))
```

## 1. Display all documents and search results

```
display(results.head(10))
```

## 2. Compare Semantic Search and Keyword Search

```
query_id = 'query01'
display(results[results['query_id'] == query_id].sort_values(['method', 'rank']))
```

## 3. Similarity Scores

```
display(scores.head(15))
```

## 4. Search Method Comparison

```
comparison = results.groupby('method').agg(
    total_results=('document_id', 'count'),
    average_score=('similarity_score', 'mean')
).reset_index()
display(comparison)
```
